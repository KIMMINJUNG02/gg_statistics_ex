# 단일 표본 t검정
- 표본 데이터가 정규성 검정
- 단일 표본 t검정
    - p-value < 0.05 : 대립가설 채택 &rightarrow; 통계적으로 유의
    - p-value >= 0.05 : 귀무가설 채택

```python
# data 불러오기
with open("./datas2/성인여성_키_데이터1.txt", "r") as f:
    data = f.read().split('\n')
    # print(data)
    # map(data_type, list_data)
    data = list(map(float, data))
    print(data)
    print(len(data))
    
```

```python
from scipy import stats
import numpy as np
```

```python
mean_value = np.mean(data)
std_dev = np.std(data, ddof=1)  # 모집단 표준편차(기본 설정)

mean_value, std_dev
```

## 가설설정
- 귀무가설 : 표본데이터의 평균은 163과 같다
- 대립가설 : 표본데이터의 평균은 163과 다르다

## 정규성 검정
- 가설설정
    - 귀무가설 : 수집 표본 데이터는 정규성을 따른다
    - 대립가설 : 수지 표본 데이터는 정규성을 따르지 않는다.

### KStest

```python
ks_statistic, p_value = stats.kstest(data, stats.norm.cdf, args=(np.mean(data), np.std(data, ddof=0)))
ks_statistic, p_value
```

```python
print(p_value)
```

- p-value > 0.05 &rightarrow; 귀무가설 채택 &rightarrow; 정규분포O

- 가설설정
    - 귀무가설 : 여성의 평균 키는 163이다.
    - 대립가설 : 여성의 평균 키는 163이 아니다.

```python
print(stats.ttest_1samp(data, 163)) 
```

- 결과해석
    - p-value < 0.05
    </br> &rightarrow; 귀무가설 기각 / 대립가설 채택
    </br> &rightarrow; 여성 평균 키 ≠ 163


# 독립 표본 t 검정

```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.read_csv("./datas2/반별_점수_type1.csv", encoding = "euc-kr")
# df1 = pd.read_csv("./datas2/반별_점수_type1.csv")
df1.head()
```

```python
groupA = df1['점수'].loc[df1['반']=='A'].values
groupB = df1['점수'].loc[df1['반']=='B'].values
print(groupA)
print(groupB)
```

```python
import numpy as np

# A반의 평균과 표준 편차
mean_A = np.mean(groupA)
std_A = np.std(groupA, ddof=1)  # 표본 표준 편차 (ddof=1)

# B반의 평균과 표준 편차
mean_B = np.mean(groupB)
std_B = np.std(groupB, ddof=1)  # 표본 표준 편차 (ddof=1)

# 결과 출력
print(f"A반 평균: {mean_A:.2f}, 표준 편차: {std_A:.2f}")
print(f"B반 평균: {mean_B:.2f}, 표준 편차: {std_B:.2f}")
```

## 가설설정
- 귀무가설 : 표본 A반과 B반의 평균은 같다.
- 대립가설 : 표본 A반과 B반의 평균은 다르다.

## 정규성 검증
- 정규성 검정 가설설정
    - 귀무가설 : 정규분포X
    - 대립가설 : 정규분포O

## 등분산성 검정

```python
from scipy import stats
```

```python
print(stats.kstest(
  groupA, 
  stats.norm.cdf, 
  args=(np.mean(groupA), np.std(groupA, ddof=1))))

print(stats.kstest(
  groupB, 
  stats.norm.cdf, 
  args=(np.mean(groupB), np.std(groupB, ddof=1))))
```

- A, B모두 p-value>0.05 &rightarrow; 두 그룹 모두 정규성을 따른다

## 등분산성 검정
- 가설 수립
    - 귀무가설 : 두 그룹은 분산이 유사하다
    - 대립가설 : 두 그룹은 분산이 유사하지 않다

```python
stats.levene(groupA, groupB) 
```

- p-value > 0.05 &rightarrow; 두 그룹의 분산은 유사하다

## 독립 표본 T검정 수행

- 가설설정
    - 귀무가설 : A와 B의 표본 평균은 같다 &mu;<sub>a</sub> =  &mu;<sub>b</sub>
    - 대립가설 : A와 B의 표본 평균은 다르다 &mu;<sub>a</sub> ≠  &mu;<sub>b</sub>

```python
# equal_var = True(등분산 만족)
print(stats.ttest_ind(groupA, groupB, equal_var = True)) 
```

- p-value < 0.05 &rightarrow; 두 반 표본평균은 다르다 &mu;<sub>a</sub> ≠  &mu;<sub>b</sub>
