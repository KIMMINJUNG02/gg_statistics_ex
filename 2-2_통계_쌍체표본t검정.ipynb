# 쌍체표본 t검정

```python
import pandas as pd
```

```python
df = pd.read_csv("datas2/다이어트약_효과검증.csv", encoding='utf-8')
df.head()
```

```python
before = df['다이어트전(kg)']
after = df['다이어트후(kg)']
```

## 정규성 검정

```python
from scipy import stats
import numpy as np
```

```python
data = after - before
```

### Shapiro-Wilk 검정
- 소표본에 강한 정규성 검정 전용 검정방식

```python
stats.shapiro(after - before)
```

- p-value > 0.05 &rightarrow; 정규분포를 따른다

### D'Agostino-Pearson 검정
- 샘플수가 n>=20일때 권장

```python
stats.normaltest(after - before)
```

- p-value > 0.05 &rightarrow; 정규분포를 따른다

## 쌍체 표본 t검정

- 가설 설정
    - 귀무가설 : 전과 후의 차이가 없다 &rightarrow; &mu;=0
    - 대립가설 : 전과 후의 차이가 있다 &rightarrow; &mu;≠0

```python
stats.ttest_rel(before, after)
```

- p-value < 0.05 &rightarrow; 전과 후의 차이가 있다 &rightarrow; 효과 있음

## 전후 차이에 대한 기술통계량

```python
import numpy as np
np.mean(before), np.mean(after)
```

```python
np.mean(before) - np.mean(after)
```

```python
# 전후 차이 계산(개인별 체중 차이)
diff = before - after

# 체중 차이의 평균 계산(평균 체중 감소량)
mean_diff = np.mean(diff)

# 체중 차이의 표준편차(감소량의 표준편차)
std_diff = np.std(diff, ddof=1)

# 체중 차이의 표준오차(표본에서 구한 평균 체중 감소량의 불확실성을 나타냄)
standard_error = std_diff / np.sqrt(len(diff))

# t-통계량 계산(평균 체중 감소량이 표준오차보다 몇 배 큰지를 나타냄)
t_statistic = mean_diff / standard_error
t_statistic
```

```python
# 계산 결과 출력
print(mean_diff)       # 평균 체중 감소량
print(std_diff)        # 체중 감소량의 표준편차
print(standard_error)  # 평균 체중 감소량의 표준오차
print(t_statistic)     # t 통계량
```

- t 통계량
    - 평균 / 표준 오차
