# 산포통계량

```python
import numpy as np
from scipy import stats
import pandas as pd
```

## 분산 & 표준편차
- `ddof` : 자유도
    - 모집단 &rightarrow; 0
    - 표본집단 &rightarrow; 1
    - default = 0
- 표준편차 : 절대적인 퍼짐

```python
x = [1, 2, 3, 4, 5]
# 분산.var
# 분모 = n - 1 ( 5 - 1), 표본집단
print(np.var(x, ddof = 1)) 
# 분모 = n, 모집단
print(np.array(x).var())
# 분모 = n , 자유도(ddof) 0:모분산인경우
print(pd.Series(x).var(ddof = 0))
```

```python
# 표준편차 .std
x = [1, 2, 3, 4, 5]
print(np.std(x, ddof = 1))
print(np.array(x).std(ddof = 0))
print(pd.Series(x).std(ddof = 1))
```

## 변동계수
- 분산과 표준편차 모두 값의 스케일에 영향을 받음
- 스케일을 동일시하기 위해 필요
- 표준편차/평균
- 의미 : *평균 대비 상대적인* 퍼짐

```python
x1 = np.array([1, 2, 3, 4, 5])
x2 = x1 * 10

print(np.std(x1, ddof = 1))
print(np.std(x2, ddof = 1))
```

```python
# 변동 계수 .variation
# 모표준편차 사용
print(stats.variation(x1)) 
print(stats.variation(x2))
```

```python
# 모표준편차 사용
print(np.std(x1, ddof = 0) / np.mean(x1))
print(np.std(x2, ddof = 0) / np.mean(x2))
```

```python
# 표본표준편차 사용
print(np.std(x1, ddof = 1) / np.mean(x1))
print(np.std(x2, ddof = 1) / np.mean(x2))
```

## 스케일링

```python
x1 = np.array([1, 2, 3, 4, 5])
x2 = x1 * 10
```

```python
# Standard Scaling
z1 = (x1 - x1.mean()) / x1.std()
z2 = (x2 - x2.mean()) / x2.std()

print(z1)
print(z2)
```

```python
# Min-max Scaling
z1 = (x1 - x1.min()) / (x1.max() - x1.min())
z2 = (x2 - x2.min()) / (x2.max() - x2.min())

print(z1)
print(z2)
```

```python
# sklearn을 이용한 스케일링을 위한 데이터 준비
X = pd.DataFrame({"X1":[1, 2, 3, 4, 5],
    "X2": [10, 20, 30, 40, 50]})

X
```

```python
# Min-max Scaling
# 0 ~ 1 데이터 스크케일링함
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler() # 인스턴스화
# 열별로 처리
Z = scaler.fit_transform(X) # fit_transform => ndarray
pd.DataFrame(Z, columns=['X1', 'X2'])
```

```python
# Standard Scaling
from sklearn.preprocessing import StandardScaler

# 표준 정규분포에 따라서 데이터의 수준을 맞춤
# 표준 정규분포 : 
ss_scaler = StandardScaler() # 인스턴스화
Z = ss_scaler.fit_transform(X) # fit_transform => ndarray
pd.DataFrame(Z, columns=['X1', 'X2'])
```
