# 대표 통계량

```python
import numpy as np
from scipy import stats
import pandas as pd
```

## 산술 평균

```python
# 성적의 평균 구하기
x = [70, 60, 80, 60, 50]
print(np.sum(x)/len(x))
print(np.mean(x))
print(np.array(x).mean())
print(pd.Series(x).mean())
```

## 조화 평균

```python
# 거리가 같을 때: 조화평균 (역수의 산술평균의 역수)
# 문제: 자동차가 60km를 60km/h로 이동하고, 같은 60km를 40km/h로 이동했을 때 평균 속도는?
x = np.array([60, 40])
print(round(len(x) / (1/60 + 1/40)))
print(round(len(x) / np.sum(1/x)))
print(round(stats.hmean(x)))
# round 반올림함수
```

- 상황: 두 생산 라인이 동일한 수의 제품(작업량)을 생산하지만, 
- 각 라인의 처리 속도가 다를 때 평균 처리 속도를 계산하기

- 라인 A: 60개/시간, 라인 B: 30개/시간  → 동일 작업량(120개 제품 생산) 처리 시 평균 처리속도

```python
# 조화평균 사용하기
x = np.array([60,30])
print(stats.hmean(x))
```

## 절사평균

```python
# 월 평균이 3백만원이고 표준편차가 50만원인 정규 분포를 따르는 소득을 갖는 100명 생성
np.random.seed(0)
income = np.random.normal(3000000, 500000, 100) 
print(round(np.mean(income), 2))
```

```python
# 월 소득이 1억원인 사람의 등장
# 100명의 소득이 300만원에 가까운데, 한 명 때문에 대표값인 평균이 약 400만원에 달함
income = np.append(income, 10**8) 
print(round(np.mean(income), 2)) 
```

```python
# 절사평균
stats.trim_mean(income, 0.2) # 하위 20%, 상위 80%를 제외하고 계산, 중앙 60%의 평균을 계산
```

## 최빈값

```python
x = np.random.choice(['A', 'B', 'C'], 1000) # A, B, C라는 요소로 데이터 1000개 만들기
print(x[:10])
print(len(x))
```

```python
np.unique(x)
```

```python
# 개수가 많은 순으로 출력
pd.Series(x).value_counts()
```

```python
pd.Series(x).value_counts().index[0] # 가장 개수가 많은 요소
```

```python
from collections import Counter
counter = Counter(x)
mode_value, mode_count = counter.most_common(1)[0]  # 최빈값과 빈도수 추출

print("최빈값:", mode_value)
print("최빈값의 빈도수:", mode_count)
```

## 데이터 대푯값

```python
#np.random.seed(0)
data = np.random.randint(0, 5, 10) # 시작, 끝, 개수
print(data)

mean = np.mean(data); print("평균값 :", mean.round(2)) # .round는 객체에 사용
median = np.median(data); print("중앙값 : ", median)
mode = stats.mode(data); print(f"최빈값 : {mode.mode}, 최빈 횟수 : {mode.count} ")
```

```python
data = pd.Series([60, 40, 60, 80, 50, 80])
print("평균값 : ", data.mean())
print("중앙값 : ", data.median())
print("최빈값 : ", data.mode())
print("최빈값 : ", data.mode().to_list())
```
